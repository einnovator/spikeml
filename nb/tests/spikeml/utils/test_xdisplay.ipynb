```python

def test_xdisplay():
    A = np.array([[0,1,3],[4,5,6]])
    xdisplay(Markup(f'{A.shape}', A))
    A = np.array([[[0,1,2],[3,4,5]], [[10,11,12],[13,14,13]]])
    xdisplay(Markup(f'{A.shape}', A))
    A = np.array([[[[0,1,2],[3,4,5]], [[10,11,12],[13,14,13]]],
                       [[[0,1,2],[3,4,5]], [[10,11,12],[13,14,13]]]])
    xdisplay(Markup(f'{A.shape}', A))

    A = np.array([
            [[[[0,1,2],[3,4,5]], [[10,11,12],[13,14,13]]],
             [[[0,1,2],[3,4,5]], [[10,11,12],[13,14,13]]]],
            [[[[0,1,2],[3,4,5]], [[10,11,12],[13,14,13]]],
             [[[0,1,2],[3,4,5]], [[10,11,12],[13,14,13]]]]])
    xdisplay(Markup(str(A.shape), A))
    xdisplay(A)

    xdisplay(Markup('T', np.array([[[0,1,2],[3,4,5]], [[10,11,12],[13,14,13]]])))
    xdisplay(Markup('T2', np.array([[[0,1,2],[3,4,5]], [[10,11,12],[13,14,13]], [[0,1,2],[3,4,5]]])))
    xdisplay(Markup('T3', np.array([[[[0,1,2],[3,4,5]], [[10,11,12],[13,14,13]]],
                       [[[0,1,2],[3,4,5]], [[10,11,12],[13,14,13]]]])))

    A = np.array([[1/3, 2, 3], [3, 4,3],[1/3, 2, 3], [3, 4,3]])
    B = np.array([[5, 6, 5], [7, 8, 5],[5, 6, 5], [7, 8, 5]])
    xdisplay(Markup('t=0', np.zeros(5), np.ones(5)), Markup('A', A), Markup('B', B),  Markup('AB', (A,B)), np.zeros(5), [1,2,3], (1,2,3), "abc", 3, 1/3)

test_xdisplay()
```
